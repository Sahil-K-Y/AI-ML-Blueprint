# Day 096 — PyTorch Autograd Engine

**Phase 6: Deep Learning & PyTorch**

## 📌 Topics
- Dynamic computational graph creation
- requires_grad=True and .backward()
- torch.no_grad() context


```python
import torch

# Create tensor and enable gradient tracking
x = torch.tensor(3.0, requires_grad=True)

# Forward pass
y = x ** 2

print("Input x:", x.item())
print("Output y:", y.item())
```

```python
y.backward()
```

```python
x.grad
```
