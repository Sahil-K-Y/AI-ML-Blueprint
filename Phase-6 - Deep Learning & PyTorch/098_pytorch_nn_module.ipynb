### 💡 Concept Overview
`nn.Module` is the fundamental base class for all neural networks in PyTorch. It automatically manages weights, biases, gradient tracking, device migration, and evaluation modes.

```python
import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Active Device: {device}")
```

### 1. Custom Neural Network Subclassing
Define layers in `__init__` and data flow in `forward()`.


```python
class MultiLayerPerceptron(nn.Module):
    def __init__(self, in_features=4, hidden_dim=16, num_classes=3):
        super().__init__()
        self.layer1 = nn.Linear(in_features, hidden_dim)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        out = self.layer1(x)
        out = self.relu(out)
        out = self.layer2(out)
        return out

model = MultiLayerPerceptron(4, 16, 3).to(device)
print(model)
```

### 2. Inspecting Trainable Parameters


```python
print("--- Trainable Parameters ---")
for name, param in model.named_parameters():
    print(f"{name:20s} | Shape: {str(list(param.shape)):15s} | Requires Grad: {param.requires_grad}")
```

### 3. Alternative: `nn.Sequential` Container


```python
seq_model = nn.Sequential(
    nn.Linear(4, 32),
    nn.ReLU(),
    nn.Linear(32, 16),
    nn.ReLU(),
    nn.Linear(16, 3)
).to(device)

dummy_x = torch.randn(5, 4, device=device)
output = seq_model(dummy_x)
print("Forward pass output shape:", output.shape)
print("Logits:\n", output)
```
