```python
from torch.utils.data import Dataset,DataLoader
```

```python
class Customdataset(Dataset):
  def __init__(self, x, y):
    self.x = torch.tensor(x, dtype=torch.float32)
    self.y = torch.tensor(y, dtype=torch.long)
  def __len__(self):
    return len(self.x)
  def __getitem__(self, idx):
    return self.x[idx], self.y[idx]
```

```python
X_train = Customdataset(X, y)
```

```python
X_test=pd.read_csv("/content/sample_data/fashion-mnist_test.csv")
```

```python
y_test=X_test.iloc[:,0]
X_test=X_test.iloc[:,1:]
```

```python
test_dataset = Customdataset(X_test.values.astype('float32') / 255.0, y_test.values)
```

```python
train_loader=DataLoader(X_train,batch_size=32,shuffle=True)
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=False)
```

```python
import torch.nn as nn

class MyNN(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.model = nn.Sequential(
        nn.Linear(num_features, 128),
        nn.ReLU(),
        nn.Linear(128, 64),
        nn.ReLU(),
        nn.Linear(64, 10)
    )
  def forward(self, x):
    return self.model(x)
```

```python
epochs=100
lr=0.1
```

```python
model = MyNN(X.shape[1])
```

```python
creterion=nn.CrossEntropyLoss()
```

```python
import torch.optim as optim

optimiser = optim.SGD(model.parameters(), lr=lr)
```

```python
for epoch in range(epochs):
  for batch_loader,batch_lebels in train_loader:
    output=model(batch_loader)
    loss=creterion(output,batch_lebels)

    optimiser.zero_grad()
    loss.backward()
    optimiser.step()
  print(f"Epoch: {epoch+1}/{epochs}, Loss: {loss.item()}")
```

```python
total = 0
correct = 0

with torch.no_grad():
  for batch_loader, batch_lebels in test_loader:
    output = model(batch_loader)
    _, predicted = torch.max(output.data, 1)
    total += batch_lebels.size(0)
    correct += (predicted == batch_lebels).sum().item()

print(f"Accuracy: {correct / total * 100:.2f}%")
```
