### 💡 Concept Overview
Weights & Biases (`wandb`) tracks code versions, hyperparameters, real-time training loss curves, and model checkpoints on interactive web dashboards.

```python
!pip install -q wandb
import wandb
import os
os.environ["WANDB_MODE"] = "offline" # Set to online when using with wandb account

config = {"lr": 0.001, "epochs": 5, "batch_size": 32, "arch": "MLP-2Layer"}
run = wandb.init(project="Phase-6-Deep-Learning", config=config)

for epoch in range(config["epochs"]):
    train_loss = 1.0 / (epoch + 1)
    acc = 75.0 + epoch * 4.0
    wandb.log({"epoch": epoch + 1, "train_loss": train_loss, "accuracy": acc})
    print(f"Epoch {epoch+1} logged to WandB.")
wandb.finish()
```
