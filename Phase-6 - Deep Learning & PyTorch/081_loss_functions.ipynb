```python
import numpy as np
```

```python
def mse_loss(y_true,y_pred):
    m=y_true.shape[0]
    return (1/(2*m))*np.sum((y_true-y_pred)**2)
```

```python
y_real = np.array([25.0, 40.0, 65.0, 80.0])
y_hat  = np.array([24.0, 42.0, 60.0, 85.0])
```

```python
mse_loss(y_real,y_hat)
```

```python
def bce_loss(y_true,y_pred,eps=1e-15):
    y_pred=np.clip(y_pred,eps,1-eps)
    m=y_true.shape[0]
    return -(1/m)*np.sum(y_true*np.log(y_pred)+(1-y_true)*np.log(1-y_pred))
```

```python
y_true_bin = np.array([[1.0], [0.0], [1.0], [0.0]])
y_pred_bin = np.array([[0.9], [0.1], [0.8], [0.2]])

bce_val = bce_loss(y_true_bin, y_pred_bin)
print("Calculated BCE Loss:", bce_val)
```

```python
def cce_loss(y_true, y_pred, eps=1e-15):
    y_pred = np.clip(y_pred, eps, 1 - eps)
    m = y_true.shape[0]
    return - (1 / m) * np.sum(y_true * np.log(y_pred))

# 1 sample: Dog is true -> [0, 1, 0]
y_true = np.array([[0, 1, 0]])
# Model guess: 80% confident on Dog
y_pred = np.array([[0.10, 0.80, 0.10]])

print("CCE Loss:", cce_loss(y_true, y_pred))
```

```python
# ============================================================
# 4. Sparse Categorical Cross-Entropy (SCCE)
# ============================================================
def scce_loss(y_true, y_pred, eps=1e-15):
    y_pred = np.clip(y_pred, eps, 1 - eps)
    m = y_true.shape[0]
    # np.arange(m) row index banata hai, y_true direct correct column pick karta hai
    correct_confidences = y_pred[np.arange(m), y_true]
    return - (1 / m) * np.sum(np.log(correct_confidences))

# Integer targets: 0 = Cat, 1 = Dog, 2 = Bird
y_sparse_true = np.array([1])  # Sample 0 is Dog (class 1)
y_sparse_pred = np.array([[0.10, 0.80, 0.10]])

scce_val = scce_loss(y_sparse_true, y_sparse_pred)
print("SCCE Loss:", scce_val)
```

```python
# ============================================================
# 5. Loss Function Selection Helper & Summary
# ============================================================
def choose_loss_and_activation(task_type, target_format="default"):
    mapping = {
        ("regression", "continuous"): ("None (Linear)", "MSE Loss"),
        ("binary_classification", "0_or_1"): ("Sigmoid", "Binary Cross-Entropy (BCE)"),
        ("multiclass_classification", "one_hot"): ("Softmax", "Categorical Cross-Entropy (CCE)"),
        ("multiclass_classification", "integer_ids"): ("Softmax", "Sparse Categorical Cross-Entropy (SCCE)"),
        ("multilabel_classification", "binary_vector"): ("Sigmoid (per class)", "BCE with Logits")
    }
    return mapping.get((task_type, target_format), ("Custom", "Custom"))

# Demonstration Table
tasks = [
    ("regression", "continuous"),
    ("binary_classification", "0_or_1"),
    ("multiclass_classification", "one_hot"),
    ("multiclass_classification", "integer_ids"),
    ("multilabel_classification", "binary_vector")
]

print("=" * 75)
print(f"{'Problem Type':<27} | {'Final Activation':<20} | {'Loss Function'}")
print("=" * 75)
for task, fmt in tasks:
    act, loss_fn = choose_loss_and_activation(task, fmt)
    print(f"{task:<27} | {act:<20} | {loss_fn}")
print("=" * 75)
```
