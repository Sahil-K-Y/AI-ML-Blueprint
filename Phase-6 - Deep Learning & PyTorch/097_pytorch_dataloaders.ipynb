# Day 097 — PyTorch Data Ingestion Architecture

**Phase 6: Deep Learning & PyTorch**

## 📌 Topics
- Custom Dataset subclassing (__len__, __getitem__)
- DataLoader (batching, shuffling, workers)
- Data preprocessing transforms


```python
import torch
from torch.utils.data import Dataset, DataLoader

# ---------------------------------------------------
# STEP 1: Custom Dataset class definition
# ---------------------------------------------------
class StudentDataset(Dataset):
    def __init__(self, marks, labels, transform=None):
        # Convert data to PyTorch tensors and store them
        self.marks = torch.tensor(marks, dtype=torch.float32)
        self.labels = torch.tensor(labels, dtype=torch.float32)
        self.transform = transform   # optional preprocessing function

    def __len__(self):
        # Return the total number of samples (called by len(dataset))
        return len(self.marks)

    def __getitem__(self, idx):
        # Retrieve a single (feature, label) pair by index
        x = self.marks[idx]
        y = self.labels[idx]

        # Apply the optional preprocessing transformation on-the-fly
        if self.transform:
            x = self.transform(x)

        return x, y


# ---------------------------------------------------
# STEP 2: Example transform function (optional step)
# ---------------------------------------------------
def normalize_marks(x):
    # Scale marks from 0-100 to a 0-1 range
    return x / 100.0


# ---------------------------------------------------
# STEP 3: Create the Dataset object
# ---------------------------------------------------
marks  = [85, 40, 90, 55, 30, 95, 60, 20]
labels = [1,  0,  1,  1,  0,  1,  1,  0]   # 1 = pass, 0 = fail

dataset = StudentDataset(marks, labels, transform=normalize_marks)

print("Total samples:", len(dataset))        # Triggers __len__
print("Sample at index 2:", dataset[2])      # Triggers __getitem__


# ---------------------------------------------------
# STEP 4: Create the DataLoader - handles batching + shuffling
# ---------------------------------------------------
dataloader = DataLoader(
    dataset,
    batch_size=3,      # Load 3 samples per batch
    shuffle=True,       # Shuffle the dataset order at every epoch
    num_workers=0        # 0 is safest and recommended for Colab
)

# ---------------------------------------------------
# STEP 5: Iterate through DataLoader batches (similar to training loop)
# ---------------------------------------------------
for batch_idx, (x_batch, y_batch) in enumerate(dataloader):
    print(f"\nBatch {batch_idx+1}")
    print("Features:", x_batch)
    print("Labels  :", y_batch)
```
