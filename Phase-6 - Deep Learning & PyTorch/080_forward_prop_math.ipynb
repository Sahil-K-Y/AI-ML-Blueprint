```python
X=np.array([[0,0],[0,1],[1,0],[1,1]])
```

```python
y=np.array([[0],[1],[1],[0]])
```

```python
def relu(z):
    return np.maximum(0,z) 
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
W1=np.random.randn(2,4)*0.1
```

```python
W1
```

```python
b1=np.zeros((1,4))
```

```python
b1
```

```python
W2 = np.random.randn(4, 1) * 0.1
b2 = np.zeros((1, 1))
```

```python
print("W1 shape:", W1.shape)
print("b1 shape:", b1.shape)
print("W2 shape:", W2.shape)
print("b2 shape:", b2.shape)
```

```python
Z1=X @ W1 +b1
```

```python
A1=relu(Z1)
```

```python
Z2=A1 @ W2 + b2
```

```python
A2=sigmoid(Z2)
```

```python
print("=" * 40)
print("FORWARD PROPAGATION OUTPUTS")
print("=" * 40)
print("Z1 shape:", Z1.shape)
print("A1 shape:", A1.shape)
print("Z2 shape:", Z2.shape)
print("A2 (Predictions) shape:", A2.shape)
print("\nFinal Output / Predictions (A2):\n", A2)
```
