# Day 058 — Hierarchical Clustering


```python
# Setup & Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

## 1. Agglomerative (Bottom-Up) vs Divisive (Top-Down) Clustering


```python
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering

X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.7, random_state=42)
X_scaled = StandardScaler().fit_transform(X)

agg = AgglomerativeClustering(n_clusters=4)
labels = agg.fit_predict(X_scaled)

plt.figure(figsize=(8, 5))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='tab10', s=30, alpha=0.7)
plt.title("Agglomerative Hierarchical Clustering (K=4)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 2. Distance Metrics & Linkage Criteria (Single, Complete, Average, Ward)


```python
linkages = ['single', 'complete', 'average', 'ward']
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, linkage_type in zip(axes.flatten(), linkages):
    model = AgglomerativeClustering(n_clusters=4, linkage=linkage_type)
    lbls = model.fit_predict(X_scaled)
    ax.scatter(X_scaled[:, 0], X_scaled[:, 1], c=lbls, cmap='viridis', s=25)
    ax.set_title(f"Linkage: {linkage_type.capitalize()}")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3. Ward's Linkage Math


```python
from scipy.cluster.hierarchy import linkage

# Compute Ward Linkage Matrix
Z_ward = linkage(X_scaled, method='ward')
print("Ward Linkage Matrix (first 5 merges):")
print(Z_ward[:5])
```

## 4. Dendrogram Plotting & Tree Cuts


```python
from scipy.cluster.hierarchy import dendrogram

plt.figure(figsize=(10, 6))
dendrogram(Z_ward, truncate_mode='lastp', p=30, leaf_rotation=90, leaf_font_size=10)
plt.axhline(y=15, color='r', linestyle='--', label='Cut Threshold (K=4)')
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Cluster Size / Leaf Index")
plt.ylabel("Distance (Merge Threshold)")
plt.legend()
plt.show()
```
